```python
import tensorflow as tf
m=tf.keras.datasets.mnist
```

```python
(x_train,y_train),(x_test,y_test)=m.load_data()
```

```python
x_train, x_test=x_train/255,x_test/255
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
])
```

```python
model.compile(optimizer='sgd',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
log=r"C:\Users\SPTINT-21\kushal\log"
from tensorflow.keras.callbacks import TensorBoard as tb
```

```python
callbacks=[tb(
    log_dir=log,
    histogram_freq=1,
    write_graph=True,
    write_images=True,
    update_freq='epoch',
    profile_batch=2,
    embeddings_freq=1
)]
```

```python
model.fit(x_train,y_train,epochs=5,validation_split=0.2,callbacks=callbacks)
model.save('ml.keras')
```
