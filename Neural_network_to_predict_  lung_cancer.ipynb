```python
import tensorflow as tf
```

```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\SPTINT-21\Desktop\survey_lung_cancer_SVM.csv")
df.columns
```

```python
x=df.drop(["LUNG_CANCER"],axis=1)
```

```python
y=df["LUNG_CANCER"]
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(512,input_dim=15,activation="relu"))
model.add(Dense(512,activation="relu"))
model.add(Dense(1,activation="sigmoid"))
    
    
```

```python
model.compile(loss="binary_crossentropy",optimizer="adam",metrics=["accuracy"])
```

```python
from tensorflow.keras.callbacks import TensorBoard
```

```python
callbacks=[TensorBoard(
        log_dir=r"C:\Users\SPTINT-21\kushal\log",
        histogram_freq=1,
        write_graph=True,
        write_images=True,
        update_freq='epoch',
        profile_batch=2,
        embeddings_freq=1
    )
]
```

```python


print("Input data types:")
print(x.dtypes)

print("Target data type:")
print(y.dtypes)

```

```python
model.fit(x,y,epochs=10,batch_size=10,callbacks=callbacks)
```

```python
_,accuracy=model.evaluate(x,y)
```

```python
print("model accuracy%.2f"%(accuracy*100))
```
