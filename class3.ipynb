```python
x=lambda a: a+5
print(x(6))
```

```python
y=lambda a,b:a+b
print(y(23,3))
```

```python
max=lambda x,y:x if x>y else y
print(max(21,3))
```

```python
square=lambda x:x*x
print(square(6))
```

```python
arr=[1,2,3,4,5,6,7]
a=list(map(lambda x:x*x,arr))
print(a)
```

```python
l1=[3,4,2,4,2]
l2=[3,5,8,2,6]
add=list(map(lambda x,y:x+y,l1,l2))
print(add)
```

```python
fruits=['mango','bannana','graps']
a=list(filter(lambda x:'g' in x,fruits))
print(a)
```
