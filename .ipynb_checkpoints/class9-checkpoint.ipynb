```python
import pandas as pd
```

```python
Data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Department": ["HR", "IT", "Finance", "Marketing"],
    "Type of Employee": ["Full-Time", "Part-Time", "Contract", "Full-Time"],
    "Salary": [50000, 35000, 45000, 55000]
}
Data
```

```python
df=pd.DataFrame(Data)
df
```

```python
import pandas as pd

PT = pd.pivot_table(
    df,
    index=['Department', 'Type of Employee'],
    values=['Salary'],
    aggfunc=['sum', 'mean','max']
)

print(PT)
```
