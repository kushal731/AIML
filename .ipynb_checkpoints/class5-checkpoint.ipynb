```python
import numpy as np
import pandas as pd
```

```python
a=np.array([1,2,3,4])
print(a)
```

```python
b=np.array([4,5,6,2])
print(b)
```

```python
ser1=pd.Series(a)
print(ser1)
```

```python
ser2=pd.Series(b)
print(ser2)
```

```python
union=pd.Series(np.union1d(ser1,ser2))
print(union)

inter=pd.Series(np.intersect1d(ser1,ser2))
print(inter)

```
