```python
import numpy as np
import pandas as pd
```

```python
a=np.array(['p','a','n','d','a','s'])
ser=pd.Series(a)
print(ser)
```

```python
x=pd.Series(4,index=[0,1,2,3])
```

```python
print(x)
```

```python
b=np.array([2,1,1,3,3,np.nan,1,1,'Luffy'])
ser=pd.Series(b)
print(ser)
```

```python
ser.value_counts()
```

```python
ser1=pd.Series([1,2,3,4,5])
ser2=pd.Series([4,5,9,6,8])
```

```python
union=pd.Series(np.union1d(ser1,ser2))
print(union)
```

```python
inter=pd.Series(np.intersect1d(ser1,ser2))
print(inter)
```
