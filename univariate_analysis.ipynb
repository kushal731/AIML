```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Dummy age list
Dummy_age_list = [12, 22, 36, 42, 15, 89, 42, 65, 29, 6,
                  35, 81, 90, 51, 53, 53, 22, 31, 75, 5]

# Dummy gender list
Dummy_sex_list = ['male', 'female', 'female', 'other', 'male',
                  'other', 'other', 'male', 'female', 'female',
                  'other', 'male', 'female', 'female', 'male',
                  'female', 'male', 'female', 'male', 'other']

# Create DataFrame
df_age_sex = pd.DataFrame({
    'AGE': Dummy_age_list,
    'Gender': Dummy_sex_list
})

# Display DataFrame
df_age_sex
```

```python
df_age_sex['AGE'].hist()
```

```python
sns.set(style='darkgrid')

# Create the count plot
ax = sns.countplot(x='Gender', data=df_age_sex,hue='Gender')
```
