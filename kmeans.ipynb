```python
import numpy as np
import matplotlib.pyplot as p
import pandas as pd
```

```python
dataset=pd.read_csv(r'C:\Users\SPTINT-21\Desktop\Mall_Customers.csv')
x=dataset.iloc[:,[3,4]].values
```

```python
from sklearn.cluster import KMeans
```

```python
wcss_list=[]
```

```python
for i in range(1,11):
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmeans.fit(x)
    wcss_list.append(kmeans.inertia_)
```

```python
p.plot(range(1,11),wcss_list)
p.title('The Elobw Method Graph')
p.xlabel('Number of cluster(k)')
p.ylabel('wcss_list')
p.show()
```

```python
kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
y_predict=kmeans.fit_predict(x)
```

```python
p.scatter(x[y_predict==0,0],x[y_predict==0,1],s=100,c='blue',label='cluster 1')
p.scatter(x[y_predict==1,0],x[y_predict==1,1],s=100,c='green',label='cluster 2')
p.scatter(x[y_predict==2,0],x[y_predict==2,1],s=100,c='red',label='cluster 3')
p.scatter(x[y_predict==3,0],x[y_predict==3,1],s=100,c='cyan',label='cluster 4')
p.scatter(x[y_predict==4,0],x[y_predict==4,1],s=100,c='magenta',label='cluster 5')
p.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],s=300,c="yellow",label='centroid')
```
