```python
q3=df['Age'].quantile(0.75)
q1=df['Age'].quantile(0.25)
IQR=q3-q1
```

```python
IQR
```

```python
lw=q1-(1.5*IQR)
```

```python
uw=q3+(1.5*IQR)
```

```python
lw
uw
```

```python
lw
```

```python
df['Age']=np.where(df['Age']<lw,lw,np.where(df['Age']>uw,uw,df['Age']))
```
