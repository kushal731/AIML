# Data Visualization on Heart Disease Dataset using Matplotlib
**CO-3 Assignment:** Histogram, Boxplot, Bar Chart, Scatter Plot, Line Graph, Pie Chart

Dataset: `heart.csv` — 1025 patient records with 14 clinical features (age, sex, cholesterol, chest pain type, max heart rate, target = presence of heart disease, etc.)

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
%matplotlib inline
```

## 2. Load Dataset

```python
df = pd.read_csv(r"C:\Users\SPTINT-21\Downloads\heart.csv")   # place heart.csv in the same folder as this notebook

print("Dataset Shape:", df.shape)
df.head()
```

```python
df.info()
```

## 3. Histogram — Age Distribution

```python
plt.figure(figsize=(8, 5))
plt.hist(df["age"], bins=15, color="#4C72B0", edgecolor="black")
plt.title("Histogram of Age Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## 4. Boxplot — Cholesterol by Heart Disease Status

```python
plt.figure(figsize=(8, 5))
data_to_plot = [df[df["target"] == 0]["chol"], df[df["target"] == 1]["chol"]]
box = plt.boxplot(data_to_plot, tick_labels=["No Disease (0)", "Disease (1)"],
                   patch_artist=True)

colors = ["#55A868", "#C44E52"]
for patch, color in zip(box["boxes"], colors):
    patch.set_facecolor(color)

plt.title("Boxplot of Cholesterol Levels by Heart Disease Status", fontsize=14, fontweight="bold")
plt.ylabel("Cholesterol (chol)")
plt.tight_layout()
plt.show()
```

## 5. Bar Chart — Patient Count by Chest Pain Type

```python
plt.figure(figsize=(8, 5))
cp_counts = df["cp"].value_counts().sort_index()
labels = ["Type 0", "Type 1", "Type 2", "Type 3"]

plt.bar(labels, cp_counts.values, color="#DD8452", edgecolor="black")
plt.title("Bar Chart: Patient Count by Chest Pain Type", fontsize=14, fontweight="bold")
plt.xlabel("Chest Pain Type (cp)")
plt.ylabel("Number of Patients")

for i, v in enumerate(cp_counts.values):
    plt.text(i, v + 5, str(v), ha="center", fontweight="bold")

plt.tight_layout()
plt.show()
```

## 6. Scatter Plot — Age vs Max Heart Rate

```python
plt.figure(figsize=(8, 5))
colors_map = df["target"].map({0: "#4C72B0", 1: "#C44E52"})

plt.scatter(df["age"], df["thalach"], c=colors_map, alpha=0.6, edgecolor="black", s=40)
plt.title("Scatter Plot: Age vs Maximum Heart Rate", fontsize=14, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Max Heart Rate (thalach)")

handles = [
    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor="#4C72B0", markersize=10, label='No Disease'),
    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor="#C44E52", markersize=10, label='Disease')
]
plt.legend(handles=handles)
plt.tight_layout()
plt.show()
```

## 7. Line Graph — Average Max Heart Rate by Age

```python
plt.figure(figsize=(8, 5))
avg_thalach_by_age = df.groupby("age")["thalach"].mean().sort_index()

plt.plot(avg_thalach_by_age.index, avg_thalach_by_age.values,
         marker="o", color="#8172B2", linewidth=2, markersize=4)
plt.title("Line Graph: Average Max Heart Rate by Age", fontsize=14, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Average Max Heart Rate")
plt.tight_layout()
plt.show()
```

## 8. Pie Chart — Gender Distribution

```python
plt.figure(figsize=(7, 7))
sex_counts = df["sex"].value_counts().sort_index()

plt.pie(sex_counts.values, labels=["Female", "Male"], autopct="%1.1f%%",
        colors=["#DD8452", "#4C72B0"], startangle=90,
        explode=(0.03, 0.03), shadow=True)
plt.title("Pie Chart: Gender Distribution of Patients", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

## Conclusion
This notebook demonstrates six core matplotlib visualization types — histogram, boxplot, bar chart, scatter plot, line graph, and pie chart — applied to the Heart Disease dataset to explore patterns in age, cholesterol, chest pain type, heart rate, and gender distribution among patients.
